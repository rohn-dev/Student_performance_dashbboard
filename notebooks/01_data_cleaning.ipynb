# Student Performance Data — Cleaning

Loads the raw dataset, standardizes column names, checks for missing values and duplicates, and writes a clean CSV for downstream analysis and the dashboard app.

## 1. Load the raw data

`DATA_PATH` points at the source CSV. Cleaning is idempotent: running this notebook again on already-clean data is safe and simply re-saves the same result.

```python
import pandas as pd
from pathlib import Path

DATA_PATH = Path("../data/study.csv")

df = pd.read_csv(DATA_PATH, encoding="latin1", on_bad_lines="skip")
print(df.shape)
df.head()
```

## 2. Inspect the data

```python
df.info()
```

```python
df.describe()
```

## 3. Missing values and duplicates

```python
print("Missing values per column:")
print(df.isnull().sum())
print("\nDuplicate rows:", df.duplicated().sum())
```

```python
# Drop any missing rows and exact duplicates
df = df.dropna()
df = df.drop_duplicates()
print(df.shape)
```

## 4. Standardize column names

Renames raw source columns (snake_case) to the consistent, human-readable labels used throughout the notebooks and the dashboard. Columns already in the target format are left untouched, so this step is safe to re-run.

```python
df = df.rename(columns={
    "gender": "Gender",
    "race_ethnicity": "Race",
    "parental_level_of_education": "Parental Level of Education",
    "lunch": "Lunch",
    "test_preparation_course": "Preparation Course",
    "math_score": "Mathematics Score",
    "reading_score": "Reading Score",
    "writing_score": "Writing Score",
})

# Drop stray 'Unnamed: 0' style index columns from earlier exports
df = df.loc[:, ~df.columns.str.contains("^Unnamed")]
df.head()
```

## 5. Final check and export

```python
df.info()
```

```python
DATA_PATH.parent.mkdir(parents=True, exist_ok=True)
df.to_csv(DATA_PATH, index=False)
print(f"Saved cleaned data to {DATA_PATH.resolve()}")
```
