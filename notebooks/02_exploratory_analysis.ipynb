# Student Performance — Exploratory Data Analysis

Visual exploration of score distributions and how they relate to demographics, parental education, and test preparation.

## Setup

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style="whitegrid")

df = pd.read_csv(Path("../data/study.csv"))
SCORES = ["Mathematics Score", "Reading Score", "Writing Score"]
print("Loaded", df.shape)
df.head()
```

## Score distributions

```python
plt.figure(figsize=(16, 4))
colors = ["tab:blue", "tab:green", "tab:red"]
for i, (col, color) in enumerate(zip(SCORES, colors), start=1):
    plt.subplot(1, 3, i)
    sns.histplot(df[col], bins=20, kde=True, color=color)
    plt.title(col)
    plt.xlabel("Score")
    plt.ylabel("Frequency" if i == 1 else "")
plt.tight_layout()
plt.show()
```

## Parental education vs. scores

```python
plt.figure(figsize=(16, 4))
for i, col in enumerate(SCORES, start=1):
    plt.subplot(1, 3, i)
    sns.boxplot(x="Parental Level of Education", y=col, data=df)
    plt.title(f"{col} by Parental Education")
    plt.xlabel("Parental Level of Education")
    plt.ylabel(col)
    plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

## Test preparation course vs. scores

```python
plt.figure(figsize=(16, 4))
for i, col in enumerate(SCORES, start=1):
    plt.subplot(1, 3, i)
    sns.boxplot(x="Preparation Course", y=col, data=df)
    plt.title(f"{col} by Preparation Course")
    plt.xlabel("Preparation Course")
    plt.ylabel(col)
plt.tight_layout()
plt.show()
```

## Gender vs. scores

```python
plt.figure(figsize=(16, 8))
for i, col in enumerate(SCORES, start=1):
    plt.subplot(2, 3, i)
    sns.boxplot(x="Gender", y=col, data=df)
    plt.title(f"{col} by Gender")
for i, col in enumerate(SCORES, start=1):
    plt.subplot(2, 3, i + 3)
    sns.violinplot(x="Gender", y=col, data=df)
    plt.title(f"{col} by Gender (distribution)")
plt.tight_layout()
plt.show()
```

## Race/ethnicity group vs. scores

```python
plt.figure(figsize=(16, 4))
for i, col in enumerate(SCORES, start=1):
    plt.subplot(1, 3, i)
    sns.boxplot(x="Race", y=col, data=df)
    plt.title(f"{col} by Race")
    plt.xlabel("Race")
    plt.ylabel(col)
plt.tight_layout()
plt.show()
```

## Correlation between scores

```python
corr = df[SCORES].corr()
plt.figure(figsize=(6, 4))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="Blues", vmin=0, vmax=1)
plt.title("Pearson Correlation between Scores")
plt.tight_layout()
plt.show()
```
